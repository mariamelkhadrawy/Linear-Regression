```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    "Age" : [1,2,3,4,5,6,7,8,] ,
    "Kilometers_Driven" : [10000,20000,30000,40000,50000,60000,70000,80000] ,
    "Price" : [900000,800000,700000,600000,500000,400000,300000,200000]
}
df = pd.DataFrame(data)
print(df)

x = df[["Age" , "Kilometers_Driven"]]
y = df["Price"]

Model = LinearRegression()
Model.fit(x,y)
print("Intercept:", Model.intercept_)
print("Coefficients:", Model.coef_)

new_Price = pd.DataFrame({
    "Age": [3],
    "Kilometers_Driven": [30000]
})

predicted_Price = Model.predict(new_Price)

print("Predicted price:", predicted_Price[0])

```
